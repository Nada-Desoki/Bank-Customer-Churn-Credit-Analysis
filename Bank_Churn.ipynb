```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('Bank Data.xlsx')
```

```python
cust = pd.read_excel('Bank Data.xlsx', sheet_name='Customer_Info')
acc = pd.read_excel('Bank Data.xlsx', sheet_name='Account_Activity')
trans = pd.read_excel('Bank Data.xlsx', sheet_name='Transaction_Credit')
```

```python
df = pd.merge(cust, acc, on='Customer_ID')
df = pd.merge(df, trans, on='Customer_ID')
```

```python
print(df.columns)
```

```python
df.drop_duplicates(inplace=True)
```

```python
df["Marital_Status"] = df["Marital_Status"].str.strip()
```

```python
df["Marital_Status"] = df["Marital_Status"].str.title()
```

```python
df["Marital_Status"] = df["Marital_Status"].replace({"Marrid":"Married", "Sngle":"Single", "Dvorced":"Divorced"})
```

```python
df.replace("Unknown", np.nan, inplace=True)

df["Age"] = df["Age"].fillna(df["Age"].median())
df["Inactive_Months_Last_Year"] = df["Inactive_Months_Last_Year"].fillna(df["Inactive_Months_Last_Year"].median())
df["Education_Level"] = df["Education_Level"].fillna(df["Education_Level"].mode()[0])
df["Marital_Status"] = df["Marital_Status"].fillna(df["Marital_Status"].mode()[0])
df["Income_Range"] = df["Income_Range"].fillna(df["Income_Range"].mode()[0])
```
